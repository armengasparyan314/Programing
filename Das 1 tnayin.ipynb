```python
number = int(input("Enter your number"))
if number > 0:
    print("This is a positive number")
elif number < 0:
    print("This is a negative number")
else:
    print("This is zero")
```

```python
number = int(input("Enter your number"))
if number % 2 == 0:
    print ("even")
else:
    print("odd")
```

```python
number1 = int(input("Enter 1 whole number"))
number2 = int(input("Enter 2 whole number"))
Type = int(input("Type 1 or 2"))
if Type == 1:
    print(number1 + number2)
elif Type ==2:
    print(number1 - number2)
else:
    print("Error")    
```
