```python
def optimize_boards(door_sizes, board_length=6):
    door_sizes = sorted(door_sizes, reverse=True)

    boards = []

    for door in door_sizes:
        placed = False

        for board in boards:
            if sum(board) + door <= board_length:
                board.append(door)
                placed = True
                break

        if not placed:
            boards.append([door])

    print("\nՏախտակների օգտագործումը:\n")

    total_waste = 0

    for i, board in enumerate(boards, start=1):
        used = sum(board)
        waste = board_length - used
        total_waste += waste

        print(f"Տախտակ {i}: {board} | օգտագործված = {used}մ | մնացորդ = {waste}մ")

    print("\nԸնդհանուր տախտակներ:", len(boards))
    print("Ընդհանուր կորուստ:", round(total_waste, 2), "մ")


# -------------------------------
# 👇 Մուտքագրում ենք տվյալները
# -------------------------------

doors = []

n = int(input("Քանի տարբեր չափի դուռ ունես? "))

for i in range(n):
    size = float(input(f"{i+1}-րդ դռան չափը (մետրով): "))
    qty = int(input(f"Քանի հատ է {size}մ դուռը? "))

    for _ in range(qty):
        doors.append(size)

optimize_boards(doors)
```

```python
# Մեր գնումների զամբյուղը
zambyugh = {
    "Koshik": [100, 10],  # Գինը՝ 100, Զեղչը՝ 10%
    "Haqust": [200, 20],  # Գինը՝ 200, Զեղչը՝ 20%
    "Gund": [300, 0]      # Գինը՝ 300, Զեղչ չկա (0%)
}

def taranjatel_ev_hashvel(gnumner):
    yndhanur_gumar = 0
    
    # .items()-ը մեզ տալիս է թե՛ ապրանքի անունը (անպետք է այս պահին), թե՛ տվյալները
    for apranq, tvyalner in gnumner.items():
        gin = tvyalner[0]
        zeghch = tvyalner[1]
        
        # Հաշվում ենք տվյալ ապրանքի վերջնական գինը
        verjnakan_gin = gin - (gin * (zeghch / 100))
        
        # Ավելացնում ենք ընդհանուր հաշվին
        yndhanur_gumar += verjnakan_gin
        
    return yndhanur_gumar

# Կանչում ենք ֆունկցիան
vcharman_taka = taranjatel_ev_hashvel(zambyugh)
print(vcharman_taka)  # 90 + 160 + 300 = 550
```
